```python
#importing the Libraies
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Reading the Dataset
dataset = pd.read_csv("Tech_Layoff_Hiring_Trends.csv")
```

```python
dataset.columns
```

```python
dataset
```

```python
dataset.isnull().sum()
```

```python
dataset.drop("record_id",axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

dataset['market_condition'] = le.fit_transform(dataset['market_condition'])

# One-Hot Encoding Columns
one_hot_cols = [
    'company_name',
    'industry',
    'country',
    'reason_for_layoffs',
    'hiring_trend',
    'top_hiring_role',
    'company_size',
    'month'
]

# One-Hot Encoding
dataset = pd.get_dummies(
    dataset,
    columns=one_hot_cols,
    drop_first=True,dtype=int
)

# Split X and y
indep_X= dataset.drop('market_condition', axis=1)
dep_Y = dataset['market_condition']

print(indep_X.shape)
print(dep_Y.shape)
print(dataset.head())
```

```python
dataset['market_condition'].value_counts()
```

```python
dataset
```

```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(indep_X, dep_Y, test_size = 1/3, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train_= sc.fit_transform(X_train)
X_test_= sc.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
#https://scikit-learn.org/stable/modules/model_evaluation.html#scoring-parameter
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'criterion':['gini','entropy'],
              'max_features': ['auto','sqrt'],
              'n_estimators':[10,100]} 



grid = GridSearchCV(RandomForestClassifier(), param_grid, refit = True, verbose = 3,n_jobs=-1,scoring='f1_macro') 
   
# fitting the model for grid search 
grid.fit(X_train_, y_train) 
 

```

```python
# print best parameter after tuning 
#print(grid.best_params_) 
re=grid.cv_results_
#print(re)
grid_predictions = grid.predict(X_test_) 
   

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, grid_predictions)



# print classification report 
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, grid_predictions)


```

```python
print(grid.best_params_)
```

```python
table=pd.DataFrame.from_dict(re)
```

```python

from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_predictions,average='weighted')
print("The f1_macro value for best parameter {}:".format(grid.best_params_),f1_macro)
```

```python
import pandas as pd
import pickle

# Load files
model = pickle.load(open("rf_model.sav", "rb"))
scaler = pickle.load(open("scaler.sav", "rb"))
columns = pickle.load(open("columns.sav", "rb"))

# User Inputs
company_name = input("Company Name: ")
industry = input("Industry: ")
country = input("Country: ")
company_size = input("Company Size: ")
month = input("Month: ")

year = float(input("Year: "))
layoffs_count = float(input("Layoffs Count: "))
layoff_percentage = float(input("Layoff Percentage: "))

reason_for_layoffs = input("Reason For Layoffs: ")

ai_automation_impact = float(input("AI Automation Impact: "))
ai_replacement_risk = float(input("AI Replacement Risk: "))
open_roles = float(input("Open Roles: "))

hiring_trend = input("Hiring Trend: ")

remote_jobs_percentage = float(input("Remote Jobs Percentage: "))

top_hiring_role = input("Top Hiring Role: ")

stock_growth_percent = float(input("Stock Growth Percent: "))
revenue_growth_percent = float(input("Revenue Growth Percent: "))
salary_budget_change = float(input("Salary Budget Change: "))

ai_adoption_level = float(input("AI Adoption Level: "))
employee_sentiment = float(input("Employee Sentiment: "))
job_security_score = float(input("Job Security Score: "))

# Create DataFrame
new_data = pd.DataFrame({
    'company_name':[company_name],
    'industry':[industry],
    'country':[country],
    'company_size':[company_size],
    'month':[month],
    'year':[year],
    'layoffs_count':[layoffs_count],
    'layoff_percentage':[layoff_percentage],
    'reason_for_layoffs':[reason_for_layoffs],
    'ai_automation_impact':[ai_automation_impact],
    'ai_replacement_risk':[ai_replacement_risk],
    'open_roles':[open_roles],
    'hiring_trend':[hiring_trend],
    'remote_jobs_percentage':[remote_jobs_percentage],
    'top_hiring_role':[top_hiring_role],
    'stock_growth_percent':[stock_growth_percent],
    'revenue_growth_percent':[revenue_growth_percent],
    'salary_budget_change':[salary_budget_change],
    'ai_adoption_level':[ai_adoption_level],
    'employee_sentiment':[employee_sentiment],
    'job_security_score':[job_security_score]
})

# One Hot Encoding
new_data = pd.get_dummies(new_data)

# Match Training Columns
new_data = new_data.reindex(columns=columns, fill_value=0)

# Scaling
new_data_scaled = scaler.transform(new_data)

# Prediction
prediction = model.predict(new_data_scaled)

print("\n Prediction : 0 - Bull Market")
print("\n              1 - Recession")
print("\n              2 - Stable")


print("\n Prediction =", prediction[0])
```
